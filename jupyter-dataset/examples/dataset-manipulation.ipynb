```python
import pandas as pd
from pathlib import Path

# Try common locations used in this project
candidate_paths = [
    Path(".sample-datasets/education/education.csv"),
    Path("education.csv"),
    Path("/opt/datasets/education/education.csv"),
]

csv_path = next((p for p in candidate_paths if p.exists()), None)
if csv_path is None:
    raise FileNotFoundError(f"Could not find education dataset in: {candidate_paths}")

df = pd.read_csv(csv_path)

print(f"Loaded: {csv_path}")
display(df.head(10))

# Quick summary of education distribution
summary = (
    df["Education"]
    .value_counts(dropna=False)
    .rename_axis("Education")
    .reset_index(name="Count")
)

display(summary)
```

```python
# Tag this cell as: dataset-transform
# (Notebook cell metadata -> tags -> add "dataset-transform")

# Example transformation:
# Keep only respondents with Bachelor's or Graduate/Professional degrees
# and standardize the education labels.
target_levels = {"Bachelor's Degree", "Graduate or Professional Degree"}

result = (
    df[df["Education"].isin(target_levels)]
    .assign(
        Education=lambda d: d["Education"].replace({
            "Bachelor's Degree": "Bachelors",
            "Graduate or Professional Degree": "Graduate"
        })
    )
    .sort_values("RespondentID")
    .reset_index(drop=True)
)

display(result.head(10))
print(f"Rows before: {len(df)}, rows after: {len(result)}")
```
